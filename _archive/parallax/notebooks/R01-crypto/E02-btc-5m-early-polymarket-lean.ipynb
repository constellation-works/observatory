## Primary τ

Outcome-signed Up mid at each frozen elapsed second, versus the uninformed 0 baseline.
Agreement is `sign(mid_up − 0.5)` versus the inferred market outcome; ties at 0.5 are dropped.


```python
import matplotlib.pyplot as plt

rows = []
for tau, frame in snapshots.items():
    rows.append(
        {
            "tau": tau,
            "mean_signed_mid": float(frame["signed_mid"].mean()),
            "agreement": float(frame.loc[~frame["tie"], "agree"].mean()),
        }
    )

fig, axes = plt.subplots(1, 2, figsize=(8, 3.2))
taus = [row["tau"] for row in rows]
axes[0].axhline(0.0, color="0.7", linewidth=1)
axes[0].plot(taus, [row["mean_signed_mid"] for row in rows], marker="o")
axes[0].set_xlabel("elapsed seconds")
axes[0].set_ylabel("mean outcome-signed mid")
axes[0].set_xticks(taus)
axes[1].axhline(0.5, color="0.7", linewidth=1)
axes[1].plot(taus, [row["agreement"] for row in rows], marker="o")
axes[1].set_xlabel("elapsed seconds")
axes[1].set_ylabel("directional agreement")
axes[1].set_xticks(taus)
fig.tight_layout()
plt.show()
```
